```python
from dotenv import load_dotenv, find_dotenv

# Load closest .env (works whether running from repo root or dev/)
load_dotenv(find_dotenv(), override=False)
```

```python
# filename: post_tweet_oauth1_tweepy.py
# Requirements:
#   pip install tweepy
#
# What this does:
# - Uses Tweepy with OAuth1 to post /2/tweets
# - Tweepy supports OAuth1; for v2 endpoints, use Client + OAuth1
#
# Note:
# - Tweepy’s v2 client can be initialized with OAuth1 credentials (bearer-only won’t work for POST /2/tweets).
import os
import tweepy

"""
Post a tweet using Tweepy with OAuth1 credentials hitting v2 /2/tweets
"""
api_key = os.getenv("X_API_KEY")
# api_key = API_KEY
api_secret = os.getenv("X_API_KEY_SECRET")
# api_secret = API_KEY_SECRET
access_token = os.getenv("X_ACCESS_TOKEN")
# access_token = ACCESS_TOKEN
access_token_secret = os.getenv("X_ACCESS_TOKEN_SECRET")
# access_token_secret = ACCESS_TOKEN_SECRET

if not all([api_key, api_secret, access_token, access_token_secret]):
    raise RuntimeError("Missing one or more env vars: X_API_KEY, X_API_SECRET, X_ACCESS_TOKEN, X_ACCESS_TOKEN_SECRET")

```

```python
# OAuth1 user context
auth = tweepy.OAuth1UserHandler(api_key, api_secret, access_token, access_token_secret)

# Tweepy v2 client with OAuth1 auth
client = tweepy.Client(
    consumer_key=api_key,
    consumer_secret=api_secret,
    access_token=access_token,
    access_token_secret=access_token_secret,
    wait_on_rate_limit=True,
    return_type=dict,  # optional: return dict instead of Tweepy Response
)
```

```python
text = "tweepy 테스트 4입니다."
```

```python
# Create tweet
response = client.create_tweet(text=text)
```

```python

# Tweepy returns Response object; convert to dict for clarity
# resp.data contains {"id": "...", "text": "..."} on success
print(f"""
    "data": {response.get("data", None)},
    "errors": {response.get("errors", None)},
    "meta": {response.get("meta", None)},
""")
```
